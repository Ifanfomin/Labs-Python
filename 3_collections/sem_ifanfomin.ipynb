## Работа со списками

Дана матрица 3х2, пример:
```
matrix = [
    [1, 2],
    [5, 6],
    [9, 10],
]
```

Необходимо написать функцию, которая будет её транспонировать, т.е. приводить к виду
```
matrix = [
    [1, 5, 9],
    [2, 6, 10],
]
```

````python
from typing import List

def transpose(matrix: List[List[int]]) -> List[List[int]]:
    hieght = len(matrix)
    width = len(matrix[0])
    new_matrix = [[0 for _ in range(hieght)] for _ in range(width)]

    for i, v1 in enumerate(matrix):
        for j, v2 in enumerate(v1):
            new_matrix[j][i] = v2

    return new_matrix


def main():
    matrix = [
        [1, 2],
        [5, 6],
        [9, 10],
    ]
    print(transpose(matrix))


if __name__ == "__main__":
    main()
````

## Работа со множествами

Даны 2 множества, к примеру

s1 = {1, 2, 3, 4, 5, 6, 7}
и s2 = {5, 6, 7, 8, 9}

Требуется:

- найти уникальные элементы из обоих множеств
- найти общие элементы для обоих множеств

````python
def get_difference(set1: set, set2: set) -> set:
    return set1 ^ set2

def get_intersecsion(set1: set, set2: set) -> set:
    return set1 & set2


def main():
    s1 = {1, 2, 3, 4, 5, 6, 7}
    s2 = {5, 6, 7, 8, 9}

    print(get_difference(s1, s2))
    print(get_intersecsion(s1, s2))


if __name__ == "__main__":
    main()
````

## Работа со словарями

Решите задачу со множествами используя словари (элементы множества - ключи словаря). 

Входные данные при этом поступают в виде списков (все элементы уникальны), к примеру:

s1 = [1, 2, 3, 4, 5, 6, 7]
и s2 = [5, 6, 7, 8, 9]

````python
from typing import List

def get_difference(list1: List[int], list2: List[int]) -> List[int]:
    dict1 = {k: "" for k in list1}
    dict2 = {k: "" for k in list2}

    return list({k: None for k in dict1 if k not in dict2}) + \
           list({k: None for k in dict2 if k not in dict1})


def get_intersection(list1: List[int], list2: List[int]) -> List[int]:
    dict1 = {k: "" for k in list1}
    dict2 = {k: "" for k in list2}

    return list({k: None for k in dict1 if k in dict2})


def main():
    s1 = [1, 2, 3, 4, 5, 6, 7]
    s2 = [5, 6, 7, 8, 9]

    print(get_difference(s1, s2))
    print(get_intersection(s1, s2))


if __name__ == "__main__":
    main()
````

#### Задача "Рекомендации"

На вход:

список recom_ids = [1, 2, 3]

список seen_ids = [1, 4, 5]

На выход:

список filtered ?= [2, 3]

Условия:
1. надо составить новый список айдишников
2. в том же порядке что recom_ids
3. НЕ содержит seen_ids

Заготовка:
```python
from typing import List

def get_filtered(recom_ids: List[int], seen_ids: List[int]) -> List[int]:
    filtered = []
    # тут логика
    return filtered
```

Требуется оценить алгоритмическую сложность полученного решения

````python
from typing import List

def get_filtered(recom_ids: List[int], seen_ids: List[int]) -> List[int]:
    filtered = []
    set_seen_ids = set(seen_ids)
    for id in recom_ids: 
        if id not in set_seen_ids:
            filtered.append(id)
    return filtered


def main():
    recom_ids = [1, 2, 3]
    seen_ids = [1, 4, 5]
    print(get_filtered(recom_ids, seen_ids))


if __name__ == "__main__":
    main()
````

алгоритмическая сложность O(n)

#### Задача "Поиск максимальных элементов в массиве"

Дан массив чисел `nums` и некоторое число `k`. Нужно написать функцию, которая достаёт `k` максимальных чисел из массива `nums`. k >= 0.

nums = [1,2,3,2,5,7,1], k = 2 -> result = [7,5]

nums = [1,2,3,3,3,3,3], k = 1 -> result = [3]

nums = [1,1], k = 1 -> result = [1]

Заготовка:
```python
from typing import List

def get_top(nums: List[int], k: int):
    top = []
    # тут логика
    return top
```

Требуется оценить алгоритмическую сложность полученного решения

````python
from typing import List

def get_top_var1(nums: List[int], k: int) -> List[int]:
    nums_copy = nums.copy()
    top = []
    max_num = [-1, -float("inf")]
    while k != 0 and len(nums_copy) != 0:
        for i, n in enumerate(nums_copy):
            if n > max_num[1]:
                max_num[0] = i
                max_num[1] = n
        nums_copy.pop(max_num[0])
        top.append(max_num[1])
        max_num = [-1, -float("inf")]
        k -= 1
        
    return top

def get_top_var2(nums: List[int], k: int) -> List[int]:
    return sorted(nums, reverse=True)[0:k]


def main():
    nums = [1,2,3,2,5,7,1]
    k = 2
    print(get_top_var1(nums, k))
    print(get_top_var2(nums, k))


if __name__ == "__main__":
    main()
````

алгоритмичаская сложность get_top_var1 O(n^2)  
алгоритмичаская сложность get_top_var2 O(nlog(n))

#### Задача "Поиск часто встречающихся элементов в массиве"

Дан массив чисел `nums` и некоторое число `k`. Нужно написать функцию, которая считает `k` самых частых чисел из массива `nums`. k >= 0.

nums = [1,2,3,2,5,7,1], k = 2 -> result = [1,2]

nums = [1,2,3,3,3,3,3], k = 1 -> result = [3]

nums = [1,1], k = 1 -> result = [1]

Заготовка:
```python
from typing import List

def get_top(nums: List[int], k: int):
    top = []
    # тут логика
    return top
```

Требуется оценить алгоритмическую сложность полученного решения

````python
from typing import List
from collections import Counter

def get_top_var1(nums: List[int], k: int) -> List[int]:
    top = []
    count = {}
    for v in nums:
        if v not in count:
            count[v] = 0
        count[v] += 1
    max_num = [0, 0]
    while k != 0 and len(count):
        for key, value in count.items():
            if value > max_num[1]:
                max_num[0] = key
                max_num[1] = value
        count.pop(max_num[0])
        top.append(max_num[0])
        max_num = [0, 0]
        k -= 1

    return top

def get_top_var2(nums: List[int], k: int) -> List[int]:
    counter = Counter(nums)
    
    sorted_items = sorted(counter.items(), key=lambda x: x[1], reverse=True)
    
    return [num for num, freq in sorted_items[:k]]

def get_top_var3(nums: List[int], k: int) -> List[int]:
    return [v[0] for v in Counter(nums).most_common(k)]

def main():
    nums = [1,2,3,2,5,7,1]
    k = 2
    print(get_top_var1(nums, k))
    print(get_top_var2(nums, k))
    print(get_top_var3(nums, k))


if __name__ == "__main__":
    main()
````

Алгоритмическая сложность get_top_var1 O(n+n^2)  
Алгоритмическая сложность get_top_var2 O(nlog(n))  
Алгоритмическая сложность get_top_var3 O(nlog(n))
