### Задание 1

Напишите функцию, которая получает список, состоящий из строчных букв (строк в 1 символ), переводит все буквы в заглавные и и соединяет все буквы в 1 слово. При решении используйте `reduce` и анонимные функции. Решение должно быть однострочником

['а', 'и', 'к', 'е'] -> 'АИКЕ'

```python
def word_maker(letters: List[str]) -> str:
    word = ...
    return word
```

Далее напишите функцию, которая сделает аналогичную работу, но без использования `reduce` и анонимных функций

````python
from typing import List
import functools

def word_maker_1(letters: List[str]) -> str:
    word = functools.reduce(lambda x, y: (x + y).upper(), letters)
    return word

def word_maker_2(letters: List[str]) -> str:
    word = ""
    for char in letters:
        word += char.upper()
    return word

print(repr(word_maker_1(['a', 'и', 'к', 'е'])))
print(repr(word_maker_2(['a', 'и', 'к', 'е'])))
````

### Задание 2

Напишите функцию, которая получает список строк и некоторое число N, и выводит в алфавитном порядке все слова (строки), состоящие из N букв.

['дом', 'улица', 'гараж', 'работа', 'он', 'она'], 3 ->  ['дом', 'она']

['дом', 'улица', 'гараж', 'работа', 'он', 'она'], 5 ->  ['гараж', 'улица']

```python
def filter_sized(words: List[str], n: int) -> List[str]:
    f = ...
    return f
```

````python
def filter_sized(words: List[str], n: int) -> List[str]:
    f = list(filter(lambda x: len(x) == n, words))
    return f

print(filter_sized(['дом', 'улица', 'гараж', 'работа', 'он', 'она'], 3))

print(filter_sized(['дом', 'улица', 'гараж', 'работа', 'он', 'она'], 5))
````

### Задание 3

Напишите функцию, которая:

    - получает на вход дату (с помощью `datetime.datetime.now()`)
    - извлекает информацию о дне, месяце и годе с помощью анонимных функций;
    - возвращает день, месяц и год
    - если  на вход приходит аргумент неверного типа, вызовите исключение
    
```python
def dmy(date: datetime.datetime) -> Tuple[int, int, int]:
    ...
    return d, m, y
```

````python
import datetime
from typing import Tuple

def dmy(date: datetime.datetime) -> Tuple[int, int, int]:
    if not isinstance(date, datetime.datetime):
        raise TypeError("date must be datetime.datetime")

    d, m, y = map(lambda f: f(date), (lambda x: x.day, lambda x: x.month, lambda x: x.year))

    return d, m, y

print(dmy(datetime.datetime.now()))
print(dmy("123"))
````

### Задание 4

Напишите функцию, которая получает на вход две строки с целыми числами и находит пересечение строк (общие для обеих строк элементы) без использования множеств. На выходе - строка. Порядок элементов не важен. Используйте `filter`

'1 2 5 7 4 8 8', '8 5 9 0' -> '8 5'

````python
def str_intersection(str1, str2) -> str:
    return ' '.join(set(filter(lambda x: x in str2 and x != ' ', str1)))

print(str_intersection('1 2 5 7 4 8 8', '8 5 9 0'))
````

### Задание 5

Напишите функцию для проверки надежности пароля. Надежный пароль должен содержать не менее N символов; кроме того, в пароле должна быть хотя бы одна:
- буква в верхнем регистре;
- буква в нижнем регистре;
- цифра;
- и хотя бы один символ из набора !@#$%^&*()-+.

'123456789i', N=5 -> False

'', N=5 -> False

'Qwerty!', N=9 -> False

'MammaMia9*', N=9 -> True

'(Abracadabrissimus5)', N=9 -> True

```python
def check_password(password: str, min_length: int) -> bool:
    is_strong = ...
    return is_strong
```

Напишите тесты на функцию, использую утверждения `assert`


````python
def check_password(password: str, min_length: int) -> bool:
    is_strong = ((len(password) >= min_length and
                 any(map(lambda x: x.isupper(), password)) and
                 any(map(lambda x: x.islower(), password)) and
                 any(map(lambda x: x.isnumeric(), password)) and
                 any(map(lambda x: x in "!@#$%^&*()-+.", password)))
                  and True) or False
    return is_strong

assert check_password('123456789i', 5) == False, "Не работает"
assert check_password('', 5) == False, "Не работает"
assert check_password('Qwerty!', 9) == False, "Не работает"
assert check_password('MammaMia9*', 9) == True, "Не работает"
assert check_password('(Abracadabrissimus5)', 9) == True, "Не работает"
````

Дополните вывод функции в случае отрицательного результата проверки пароля списком замечаний

'123456789i', N=5 -> False, ['заглавная', 'символ']

'', N=5 -> False, ['заглавная', 'символ', 'прописная', 'цифра', 'длина 0 символов из 5']

````python
def check_password(password: str, min_length: int):
    errors = []

    has_upper = any(map(lambda x: x.isupper(), password))
    has_lower = any(map(lambda x: x.islower(), password))
    has_digit = any(map(lambda x: x.isnumeric(), password))
    has_symbol = any(map(lambda x: x in "!@#$%^&*()-+.", password))

    if not has_upper:
        errors.append('заглавная')

    if not has_symbol:
        errors.append('символ')

    if not has_lower:
        errors.append('прописная')

    if not has_digit:
        errors.append('цифра')

    if len(password) < min_length:
        errors.append(f'длина {len(password)} символов из {min_length}')

    if errors:
        return False, errors

    return True

assert check_password('123456789i', 5) == (False, ['заглавная', 'символ']), "Не работает"
assert check_password('', 5) == (False, ['заглавная', 'символ', 'прописная', 'цифра', 'длина 0 символов из 5']), "Не работает"
assert check_password('MammaMia9*', 9) == True, "Не работает"
assert check_password('(Abracadabrissimus5)', 9) == True, "Не работает"
````
