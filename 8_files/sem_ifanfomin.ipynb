### Задача 1

Анализ лог-файла

Дан access-лог nginx access_log, в котором N строк. 
Нужно найти в нём:
- 10 User-Agent-ов с которых пришло больше всего запросов
- 10 IP на которые ушло больше всего трафика в сутки (пиковое значение) 

Оценить сложность работы предложенного алгоритма.

Формат access-лога стандартный: один запрос - одна строка, в строке поля разделены пробелом, IP в самом первом поле.

Пример:
```
ip - - date request response_code size - user_agent

93.180.71.3 - - [17/May/2015:08:05:32 +0000] "GET /downloads/product_1 HTTP/1.1" 304 0 "-" "Debian APT-HTTP/1.3 (0.8.16~exp12ubuntu10.21)"
```


````python
from typing import NamedTuple
from collections import Counter
from pprint import pprint

class Info(NamedTuple):
    ip: str
    date: str
    response_code: int
    size: int
    user_agent: str

infos = []
with open("access_log", "r") as file:
    for line in file:
        ip = line.split(" ")[0]
        date = line.split("[")[-1].split("]")[0]
        request = line.split("\"")[1]
        response_code = line.split("\"")[2].split(" ")[1]
        size = line.split("\"")[2].split(" ")[2]
        user_agent = line.split("\"")[-2]
        infos.append(Info(ip, date, int(response_code), int(size), user_agent))


# O(n + nlogn)
users_agents_counter = Counter(info.user_agent for info in infos) 
pprint(users_agents_counter.most_common(10))

# O(n + nlogn)
traffic = {}
for info in infos:
    if info.ip not in traffic:
        traffic[info.ip] = 0

    traffic[info.ip] += info.size

traffic = sorted(traffic.items(), key=lambda x: x[1], reverse=True)
print()
pprint(traffic[:10])
````
