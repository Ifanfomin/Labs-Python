Сервер (и клиент) помимо обычных сообщений от клиента (сервера) должен принимать специальные команды (зарезервированные слова), например:

### Задача 2

Воспользуйтесь публичным [api hh.ru](https://github.com/hhru/api#%D1%81%D0%BF%D1%80%D0%B0%D0%B2%D0%BE%D1%87%D0%BD%D0%B8%D0%BA%D0%B8). Используя **доступные** данные обо всех станциях метро (metro) и регионах (areas), определите, в каких странах сколько станций метро

```python
import requests
from collections import defaultdict

def get_countries():
    response = requests.get('https://api.hh.ru/areas')
    return response.json()

def get_metro():
    response = requests.get('https://api.hh.ru/metro/')
    return response.json()

metro_list = get_metro()
countries_list = get_countries()

stations_count = defaultdict(int)

for metro in metro_list:
    city_id = metro["id"]
    for country in countries_list:
        for city in country["areas"]:
            if city_id == city["id"]:
                stations_count[country["name"]] += sum(len(line["stations"]) for line in metro["lines"])

print(stations_count.items())
```
