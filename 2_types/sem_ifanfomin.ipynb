## Ввод и вывод

#### Задание 1 (Калькулятор возраста питомца)

**На вход:** получить из ввода (input) две строки. В первой — имя питомца (например, "Бобик"), во второй — возраст питомца в годах (целое число).

**На выходе:** одна строка, в которой сообщается, сколько примерно месяцев питомцу.

Формат вывода: `Бобику 24 месяца, или 2 года и 0 месяцев.`

````python
import pymorphy3

morph = pymorphy3.MorphAnalyzer()

pet_name = str(input("Введи имя питомца: "))
pet_name = morph.parse(pet_name)[0]
pet_age_years = int(input("Введи возраст питомца в годах (целое число): "))
pet_age_month = pet_age_years * 12

month_str = ""
if pet_age_month % 100 in [11, 12, 13, 14] or pet_age_month % 10 in [0, 5, 6, 7, 8, 9]:
    month_str = "месяцев"
elif pet_age_month % 10 == 1:
    month_str = "месяц"
elif pet_age_month % 10 in [2, 3, 4]:
    month_str = "месяца"

years_str = ""
if pet_age_years % 100 in [11, 12, 13, 14] or pet_age_years % 10 in [0, 5, 6, 7, 8, 9]:
    years_str = "лет"
elif pet_age_years % 10 == 1:
    years_str = "год"
elif pet_age_years % 10 in [2, 3, 4]:
    years_str = "года"


print(f"{pet_name.inflect({"datv"}).word.capitalize()} {pet_age_month} {month_str}, или {pet_age_years} {years_str}")
````

#### Задание 2 (Реверс строки)

**На вход:** получить из ввода строку из трёх слов, разделённых пробелами (например, "в лесу родилась")

**На выходе:** строка, где слова идут в обратном порядке и разделены подчеркиваниями.

Предложите несколько способов соединения слов

````python
words = str(input("Введите три слова: "))

# 1 способ
print("_".join(words.split()[::-1]))

# 2 способ
word = ""
output = ""
for char in words[::-1]:
    if char != " ":
        word = char + word
    else:
        output = output + word + "_"
        word = ""
else:
    output = output + word

print(output)
````

## Арифметические операции

#### Задание 1 (Чаевые в ресторане)

**На вход:** сумма счёта (целое число, например, 2350) и процент чаевых, который вы хотите оставить (целое число, например, 15).

**На выходе:** итоговая сумма к оплате (счёт + чаевые) и сумма чаевых в рублях.

Формат вывода: `Счёт: 2350 руб. Чаевые (15%): 352.5 руб. Итого: 2702.5 руб.`

````python
score = int(input("Сумма счета (целое число): "))
percent = int(input("Процент чаевых (целое): "))
percent_score = score * percent / 100

print(f"Счёт: {score} руб. Чаевые ({percent}%): {percent_score} руб. Итого: {score + percent_score}")
````

#### Задание 2 (Площадь кольца)

**На вход:** два целых числа: внешний радиус (R) и внутренний радиус (r) кольца.

**На выходе:** строка с данными и результатом вычисления площади кольца (S = π*(R² - r²)).

Формат вывода: `Площадь кольца с радиусами 5 и 3 = 50.24`

````python
from math import pi

R = float(input("Внешний радиус: "))
r = float(input("Внутренний радиус: "))

print(f"Площадь кольца с радиусами {R} и {r} = {pi * (R**2 - r**2)}")
````

#### Задание 1 (Пропуск на мероприятие 18+)

**На вход:** получить из ввода три значения через input() (каждое на отдельной строке):

    Возраст посетителя (целое число).

    Наличие паспорта (строка "да" или "нет").

    Наличие пригласительного (строка "да" или "нет").

**На выходе:** Строка 'Проходите' или 'Вход воспрещён'.

**Условия:** Пройти можно, если:

    Возраст 18 или больше И есть паспорт.

    ИЛИ есть пригласительный (пригласительный отменяет требование возраста и паспорта).


````python
age = int(input("Возраст посетителя (целое число): "))

passport = str(input("Наличие паспорта (строка \"да\" или \"нет\"): "))
assert passport.lower() in ["да", "нет"], "Укажите ДА или НЕТ насчет паспорта"
passport = True if passport == "да" else False

invitation = str(input("Наличие пригласительного (строка \"да\" или \"нет\"): "))
assert invitation.lower() in ["да", "нет"], "Укажите ДА или НЕТ насчет паспорта"
invitation = True if invitation == "да" else False


if age >= 18 and passport or invitation:
    print("Проходите")
else:
    print("Вход воспрещён")
````

#### Задание 1 (Генератор никнеймов)

Пользователь вводит имя и фамилию через пробел. Программа генерирует никнейм.

    Первые 3 буквы фамилии (в верхнем регистре).

    Символ подчеркивания.

    Первые 2 буквы имени (в нижнем регистре).

    Длина полного имени (количество символов, включая пробел).

**Пример:**

    Ввод: Марина Александрова
    
    Вывод: 'АЛЕ_ма10'
    
    Пояснение: АЛЕ (первые 3 от Александрова), ма (первые 2 от Марины в нижнем), 10 (в строке "Марина Александрова" 10 символов, считая пробел).

````python
full_name = str(input("Введите имя и фамилию через пробел"))
assert len(full_name.split()) == 2, "Просто ИМЯ и ФАМИЛИЯ через ОДИН пробел"

first, second = full_name.split()
nickname = f"{second[:3].upper()}_{first[:2].lower()}{len(full_name)}"

print(nickname)
````

#### Задание 2 ("Угадай число")

Программа "загадывает" число от 1 до 100 (можно использовать переменную, например secret = 42).
Пользователю дается 5 попыток, чтобы угадать.

**На вход:** В цикле вводятся числа.

**На выходе:** После каждой попытки программа подсказывает "Больше" или "Меньше". Если число угадано — выводится "Поздравляю!" и цикл прерывается. Если попытки закончились — выводится "Попытки кончились. Было загадано: 42".

````python
import random

secret = random.randint(1, 100)

print("Я загадал число от 1 до 100")
print("Попробуй угадать за 5 попыток")
for i in range(5):
    print(f"Попытка {i + 1} из 5")
    num = int(input("Угадай число: "))
    print(f"Пользователь ввёл: {num}")
    if secret > num:
        print("Больше")
    elif secret < num:
        print("Меньше")
    else:
        print("Поздравляю!")
        break
else:
    print(f"Попытки кончились. Было загадано: {secret}")
````

####  Задание 3 (Сумма цифр числа)

**На вход:** целое неотрицательное число.

**На выходе:** сумма его цифр.

**Условие:** Строки использовать нельзя.

**Пример:** 

    Ввод: 1234
    
    Вывод: 10

````python
num = int(input("Целое неотрицательное число: "))
print(sum(map(int, str(num))))
````
